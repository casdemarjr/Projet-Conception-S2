#Projet Conception

Ce projet a pour objectif d'optimiser l'acoustique interne de l'amphithéâtre E. Freyssinet (département GCU) en menant conjointement l'étude d'installation d'une nouvelle Centrale de Traitement de l'Air (CTA). À l'aide du logiciel de prévision architecturale CATT-Acoustic et des données de base fournies (dimensions sur plan, coefficients d'absorption Sabine), il s'agira de définir les matériaux de revêtement adéquats tout en évaluant l'impact sonore du système de ventilation. L'enjeu principal est d'analyser l'interaction directe entre ces deux éléments, notamment en quantifiant les nuisances générées par les bouches de soufflage de la CTA, afin de garantir et de préserver une qualité d'écoute optimale au sein de la salle.


```python
#Données
n = 100 #nombre étudiants
rho = 1.204 #kg/m3
debitairneuf = 30 #m3/h par étudiants
mairneuf = rho*debitairneuf #Débit massique d'air neuf
debiti = 300 #m3/h
mi = rho*debiti #Débit massique d'infiltration
LED = 1400 #W éclairage led
U = 0.3 #W/m2/K murs
Smurs = 6.5*14
Splafond = 14*12
```

##Climatisation été



```python
#Données été
Te = 30 #°C
Ti = 24 #°C
HRe = 70 #%
HRi = 50 #%
re = 0.0188 #kg/kgas
ri = 0.0933 #kg/kgas
Qsetu = 67 #W par étudiants
Qvetu = 60 #g/h par étudiants
psirouge = 0.67
psijaune = 1.36
psibleu = 0.76
L = 12
cp = 1000

#Charges sensibles
Qetudiants = n*Qsetu
Qmurs = U*(Smurs+Splafond)*(Te-Ti)
phirouge = psirouge * L *(Te-Ti)
phijaune = psijaune * L *(Te-Ti)
Qi = (mi/3600)*cp*(Te-Ti)
Qled = 1400

Qszt = Qetudiants+Qmurs+phirouge+phijaune+Qi+Qled     #à vérifier les signes
print (Qszt)



```
