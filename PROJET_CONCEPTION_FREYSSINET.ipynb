#Hypothèses de calcul

##**1. Hypothèses Météorologiques (Données Extérieures)**

Les conditions extérieures sont définies pour les deux situations extrêmes afin de dimensionner les installations de manière sécuritaire.

Période Hivernale : Température de base extérieure ($\theta_e$) : $-8\text{°C}$.  Humidité relative ($HR_e$) : $40\%$ (air froid et saturé).  
Période Estivale : Température de base extérieure ($\theta_e$) : $+30\text{°C}$.  Humidité relative ($HR_e$) : $70\%$.  


##**2. Hypothèses d'Occupation et Apports Internes**

L'amphithéâtre est caractérisé par une forte densité d'occupation, ce qui influence majoritairement le bilan thermique.

Capacité d'accueil : 100 étudiants (considérés comme "spectateurs assis" pour le métabolisme).  On trouve donc les apports métaboliques unitaires selon les données Carrier :

En Hiver (à 20°C) : Chaleur sensible = $60\text{ W/pers}$ | Chaleur latente = $42\text{ W/pers}$.  

En Été (à 24°C) : Chaleur sensible = $67\text{ W/pers}$ | Chaleur latente = $35\text{ W/pers}$.  

Éclairage : Puissance installée de $1400\text{ W}$ (technologie LED), considérée comme un apport de chaleur sensible constant lors de l'occupation.

##**3. Hypothèses de Confort (Données Intérieures)**

Les conditions intérieures sont fixées pour répondre aux exigences de confort thermique et d'hygiène des occupants.

Températures de consigne (intérieures) ($\theta_i$) :

Hiver : $20\text{°C}$.  
Été : $24\text{°C}$.

L'humidité de confort est maintenue à environ $50\%$ d'humidité relative.

Qualité de l'air (Hygiène) : Débit d'air neuf réglementaire : $30\text{ m}^3\text{/h}$ par personne.  

Débit d'air neuf total minimal : $3000\text{ m}^3\text{/h}$.  

Étanchéité du bâti : Un débit d'infiltrations d'air parasite de $300\text{ m}^3\text{/h}$ est pris en compte dans les calculs de déperditions.  

##**4. Hypothèses sur l'Enveloppe du Bâtiment**

Le calcul des charges sensibles repose sur les caractéristiques thermiques des parois déperditives identifiées :  
Coefficient de transmission thermique ($U$) : $0,3\text{ W/m}^2\text{.K}$ pour le mur Ouest et la toiture terrasse (plafond).  

Ponts thermiques linéiques (Façade Ouest) :

Liaison Mur/Sol : $\psi = 1,36\text{ W/m.K}$ et Liaison Mur/Plafond : $\psi = 0,67\text{ W/m.K}$.  
Parois neutres : Les murs Nord, Sud et Est sont considérés comme adjacents à des locaux chauffés à la même température, n'engendrant ainsi aucun flux thermique.

```python
#Données
n = 100 #nombre étudiants
rho = 1.204 #kg/m3
debitairneuf = 30 #m3/h par étudiants
mairneuf = rho*debitairneuf #Débit massique d'air neuf
debiti = 300 #m3/h
mi = rho*debiti #Débit massique d'infiltration
LED = 1400 #W éclairage led
U = 0.3 #W/m2/K murs
Smurs = 6.5*12
Splafond = 14*12
Lv=2496000 #J/kg
```

#Charges hiver et été



##Climatisation été

Dans cette section, nous évaluons les charges thermiques sensibles et latentes de l'amphithéâtre en régime d'été (salle occupée) afin de pouvoir dimensionner les équipements de traitement d'air.

**Données climatiques et internes de base :**
* **Conditions extérieures (Lyon) :** $T_e$ = **30°C** et $HR_e$ = **70%** (soit une humidité absolue $r_e \approx$ **18,8 g/kgas**).
* **Conditions intérieures visées :** $T_i$ = **24°C** et $HR_i$ = **50%** (soit $r_i \approx$ **9,3 g/kgas**).
* **Occupation (100 étudiants) :** D'après les données constructeurs pour un spectateur à **24°C**, chaque occupant dissipe une chaleur sensible de **67 W** et un apport d'eau de **72 g/h**.
* **Apports internes divers :** Éclairage LED de **1400 W**.

**Démarche de calcul :**

-  **Charges Sensibles ($Q_{szt}$) :** Elles regroupent l'ensemble des apports de chaleur pure faisant grimper la température de la pièce. Le bilan fait la somme :
    - Des apports internes (étudiants et éclairage).
    - Des transferts par conduction à travers les parois déperditives (mur Ouest et plafond), proportionnels au coefficient $U$ et à l'écart de température ($T_e - T_i$).
    - Des transferts par les ponts thermiques linéiques (coefficients $\psi_{rouge}$ et $\psi_{jaune}$).
    - De la charge sensible liée aux infiltrations d'air extérieur chaud (**300 m³/h**).
-  **Charges Latentes ($Q_{lzt}$) :** Elles correspondent aux apports d'humidité qu'il faudra condenser (déshumidifier). Le bilan somme :
    * L'humidité générée par la respiration et la sudation des occupants (convertie en puissance via la chaleur latente de vaporisation de l'eau $L_v$).
    * La charge latente liée aux infiltrations, l'air extérieur étant plus chargé en humidité que l'air intérieur en été.


```python
#Données été
Te = 30 #°C
Ti = 24 #°C
HRe = 70 #%
HRi = 50 #%
re = 0.0188 #kg/kgas
ri = 0.00933 #kg/kgas
Qsetu = 67 #W par étudiants
Qvetu = 35 #W par étudiants
psirouge = 0.67
psijaune = 1.36
psibleu = 0.76
L = 12
cp = 1000


#Charges sensibles
Qetudiants = n*Qsetu
Qmurs = U*(Smurs+Splafond)*(Te-Ti)
phirouge = psirouge * L *(Te-Ti)
phijaune = psijaune * L *(Te-Ti)
Qi = (mi/3600)*cp*(Te-Ti)
Qled = 1400

Qszte = Qetudiants+Qmurs+phirouge+phijaune+Qi+Qled
print (Qszte)


#Charges latentes
Qetudiants = n*Qvetu
Qi = mi*Lv*(re-ri)/3600

Qlzte = Qetudiants + Qi
print(Qlzte)
```

##Climatisation Hiver

Dans cette section, nous évaluons les charges thermiques sensibles et latentes de l'amphithéâtre en régime d'hiver (salle occupée). L'enjeu de ce calcul est de vérifier l'équilibre entre les déperditions par l'enveloppe et les importants apports internes liés à la présence de 100 personnes.

**Données climatiques et internes de base :**
* **Conditions extérieures (Lyon) :** $T_e$ = **-8°C** et $HR_e$ = **40%** (soit une humidité absolue très faible $r_e =$ **0,0008 kg/kgas**).
* **Conditions intérieures visées :** $T_i$ = **20°C** et $HR_i$ = **50%** (soit $r_i =$ **0,0073 kg/kgas**).
* **Occupation (100 étudiants) :** D'après les données constructeurs à **20°C**, chaque occupant dissipe une chaleur sensible de **60 W** et génère un apport d'eau de **60 g/h**.
* **Apports internes divers :** Éclairage LED de **1400 W**.

**Démarche de calcul :**

-  **Charges Sensibles ($Q_{szt}$) :** Le bilan thermique est une somme algébrique qui oppose :
    * Les **déperditions thermiques** vers l'extérieur : transferts par conduction (mur Ouest et plafond), ponts thermiques, et pertes par infiltrations d'air froid. En utilisant le terme ($T_e - T_i$), ces valeurs sont structurellement négatives.
    * Les **apports internes gratuits** : métabolisme des étudiants et éclairage (valeurs positives).
-  **Charges Latentes ($Q_{lzt}$) :** Le bilan hydrique prend en compte :
    * La vapeur d'eau générée par la respiration des occupants (gain, positif).
    * La perte d'humidité due aux infiltrations (l'air à -8°C entrant dans la pièce étant beaucoup plus sec que l'air intérieur, il crée un "assèchement", ce qui représente une charge latente négative).

```python
#Données hiver
Te = -8 #°C
Ti = 20 #°C
HRe = 40 #%
HRi = 50 #%
re = 0.0008 #kg/kgas
ri = 0.0073
Qsetu = 60 #W par étudiants
Qvetu = 42 #W par étudiants
psirouge = 0.67
psijaune = 1.36
psibleu = 0.76
L = 12
cp = 1000


#Charges sensibles
Qetudiants = n*Qsetu
Qmurs = U*(Smurs+Splafond)*(Te-Ti)
phirouge = psirouge * L *(Te-Ti)
phijaune = psijaune * L *(Te-Ti)
Qi = (mi/3600)*cp*(Te-Ti)
Qled = 1400

Qszth = Qetudiants+Qmurs+phirouge+phijaune+Qi+Qled
print (Qszth)


#Charges latentes
Qetudiants = n*Qvetu
Qi = mi*Lv*(re-ri)/3600

Qlzth = Qetudiants + Qi
print(Qlzth)
```

# Débit de soufflage

Une fois les charges thermiques du local établies, l'étape suivante consiste à dimensionner le débit d'air à souffler par la Centrale de Traitement d'Air (CTA) et à déterminer sa température d'injection.

Pour garantir le confort, la qualité de l'air et compenser les charges, le débit de soufflage retenu doit satisfaire simultanément **trois conditions**. Le débit final sera le maximum de ces trois valeurs.

**Démarche de calcul :**

1.  **Le critère thermique ($m_1$) :** Le débit doit être suffisant pour vaincre les charges sensibles du local ($Q_{sh}$ en hiver, $Q_{se}$ en été) avec un écart de température de soufflage initialement fixé (ici $\Delta T =$ **14°C**).
2.  **Le critère hygiénique ($m_2$) :** Le renouvellement d'air neuf (réglementaire) s'impose pour garantir la qualité de l'air. [cite_start]Pour l'amphithéâtre, il est fixé à **30 m³/h par étudiant** [cite: 96] [cite_start]pour **100 étudiants**[cite: 95].
3.  **Le critère de brassage ($m_3$) :** Pour assurer une bonne homogénéité des températures et une diffusion correcte de l'air dans le grand volume de l'amphithéâtre ($V \approx$ **1092 m³**), on impose un taux de brassage minimum : **5 vol/h** en hiver et **8 vol/h** en été.

**Recalcul des températures de soufflage réelles :**
Puisque l'on conserve le débit le plus contraignant (le maximum des trois), le débit réellement soufflé sera souvent bien supérieur au besoin purement thermique. Par conséquent, il est nécessaire de recalculer la température de soufflage réelle ($T_{sh}$ et $T_{se}$) pour ne pas sur-refroidir ou surchauffer la pièce.

```python
Tih = 20
Tie = 24
dT = 14
Tsh = Tih + dT
Tse = Tie - dT
Qsh = Qszth
Qse = Qszte
Vamphi = 12*14*6.5
rhoair= 1.2
netu = 100
metu = 30
bh = 5
be = 8
m1h = Qsh/(1000*dT)
m1e = Qse/(1000*dT)
m2 = netu*metu*rhoair/3600
m3h = Vamphi*bh*rhoair/3600
m3e = Vamphi*be*rhoair/3600
mh = max(m1h,m2,m3h)
me = max(m1e,m2,m3e)
print(mh,me)
Tsh = Tih-Qsh/(mh*1000)
Tse = Tie-Qse/(me*1000)
print(Tsh,Tse)
```

## Dimensionnement de la CTA

Conformément aux exigences du projet, nous allons modéliser la Centrale de Traitement d'Air (CTA) en intégrant un **caisson de mélange** (recyclage d'une partie de l'air extrait) et un **échangeur à plaques** pour récupérer l'énergie sur l'air rejeté.

### Démarche Thermodynamique :

* **Échangeur à plaques :** L'air neuf hygiénique croise l'air vicié extrait pour se préchauffer (en hiver) ou se pré-refroidir (en été) sans que les flux ne se mélangent. Nous adoptons une efficacité thermique $\eta = 70\%$ (valeur standard cohérente avec la documentation technique du projet).


* **Caisson de mélange :** Le débit total de soufflage ($m_h$ ou $m_e$) étant très supérieur au débit d'air neuf hygiénique ($m_2$), la CTA recycle une grande partie de l'air ambiant. L'air neuf (prétraité par l'échangeur) se mélange donc à l'air recyclé (aux conditions de confort de la pièce) avant d'entrer dans les sections suivantes.


* **Batteries thermiques :** La puissance sensible des batteries est calculée pour amener le flux d'air depuis sa température de mélange ($T_{\text{mix}}$) jusqu'à la température de soufflage réelle requise ($T_{\text{sh}}$ ou $T_{\text{se}}$) afin de compenser exactement les charges du local.


* **Section d'Humidification (Régime Hiver) :** L'air extérieur hivernal étant extrêmement sec ($r_e = 0,8\text{ g/kg}_{as}$), son introduction et son réchauffement provoquent un effet d'assèchement majeur au sein de l'amphithéâtre. Afin de maintenir fermement la consigne de confort de $50\%$ d'humidité relative, une rampe d'humidification par injection de vapeur est implantée en aval de la batterie chaude. Le débit d'eau nécessaire est déterminé par un bilan hydrique global du local, confrontant le déficit d'humidité lié au renouvellement d'air et aux infiltrations parasites avec les apports de vapeur endogènes des étudiants.

L'enjeu de cette étude est de **dimensionner la puissance thermique des batteries**, de **définir la capacité nominale de l'humidificateur d'hiver**, et de **quantifier précisément l'économie d'énergie** réalisée grâce à la présence de l'échangeur à plaques.

```python
eff_echangeur = 0.70
Te_h = -8
Te_e = 30
cp = 1000

T_airneuf_h = Te_h + eff_echangeur * (Tih - Te_h)
m_recyc_h = mh - m2
T_mix_h = (m2 * T_airneuf_h + m_recyc_h * Tih) / mh
P_chaude_avec_echangeur = mh * cp * (Tsh - T_mix_h)
T_mix_h_sans = (m2 * Te_h + m_recyc_h * Tih) / mh
P_chaude_sans_echangeur = mh * cp * (Tsh - T_mix_h_sans)

economie_hiver = P_chaude_sans_echangeur - P_chaude_avec_echangeur

perte_vapeur_air_neuf = (m2 * 3600) * (ri - re)
perte_vapeur_infilt = mi * (ri - re)
gains_vapeur_etudiants = n * (Qvetu / 1000)
debit_humidificateur = perte_vapeur_air_neuf + perte_vapeur_infilt - gains_vapeur_etudiants

T_airneuf_e = Te_e - eff_echangeur * (Te_e - Tie)
m_recyc_e = me - m2
T_mix_e = (m2 * T_airneuf_e + m_recyc_e * Tie) / me
P_froide_avec_echangeur = me * cp * (T_mix_e - Tse)
T_mix_e_sans = (m2 * Te_e + m_recyc_e * Tie) / me
P_froide_sans_echangeur = me * cp * (T_mix_e_sans - Tse)

economie_ete = P_froide_sans_echangeur - P_froide_avec_echangeur


# RÉSULTATS

print("BILAN CTA HIVER")
print(f"Temp. air neuf après échangeur : {T_airneuf_h:.1f} °C")
print(f"Temp. de mélange dans la CTA   : {T_mix_h:.1f} °C")
print(f"Temp. cible de soufflage       : {Tsh:.1f} °C")
print(f"-> Puissance Batterie Chaude AVEC échangeur : {P_chaude_avec_echangeur/1000:.1f} kW")
print(f"-> Puissance Batterie Chaude SANS échangeur : {P_chaude_sans_echangeur/1000:.1f} kW")
print(f"ÉCONOMIE DE CHAUFFAGE RÉALISÉE : {economie_hiver/1000:.1f} kW")
print(f"--> DÉBIT D'EAU REQUIS POUR L'HUMIDIFICATEUR : {debit_humidificateur:.2f} kg/h\n")

print("BILAN CTA ÉTÉ (Puissance Sensible)")
print(f"Temp. air neuf après échangeur : {T_airneuf_e:.1f} °C")
print(f"Temp. de mélange dans la CTA   : {T_mix_e:.1f} °C")
print(f"Temp. cible de soufflage       : {Tse:.1f} °C")
print(f"-> Puissance Batterie Froide AVEC échangeur : {P_froide_avec_echangeur/1000:.1f} kW")
print(f"-> Puissance Batterie Froide SANS échangeur : {P_froide_sans_echangeur/1000:.1f} kW")
print(f"ÉCONOMIE DE REFROIDISSEMENT RÉALISÉE : {economie_ete/1000:.1f} kW")
```
